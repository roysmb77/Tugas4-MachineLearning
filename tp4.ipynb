```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Online Sales Data.csv")

# Konversi kolom tanggal menjadi tipe datetime dan set sebagai index
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# Membuat plot garis dari kolom target
plt.figure(figsize=(12, 6))
df['Total Revenue'].plot()
plt.title('Grafik Total Pendapatan Harian Penjualan Online')
plt.xlabel('Tanggal')
plt.ylabel('Total Pendapatan (USD)')
plt.grid(True)
plt.show()
```
